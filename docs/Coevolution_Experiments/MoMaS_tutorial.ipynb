#### Load Configuration Parameters 

```python
params = MaaSSim.utils.get_config('glance.json')
```

#### Parameters: Configure the Simulation City

```python
params.city = "Amsterdam, Netherlands"
```

#### Parameters: Set the Path for City Graph and Distance Matrix (Skim)

```python
params.paths.G = "../../data/graphs/Amsterdam.graphml"
params.paths.skim = "../../data/graphs/Amsterdam.csv"
```

#### Parameters: Set True if You Want Drivers (Vehicles) to Start from the Same Position Every Day

```python
params.vehicle_fixed_positions = True
```

#### Parameters: Set Reserwation Wage for Drivers

```python
params.d2d.res_wage = 42.52 # in Euro for the one day simulation period
```

#### Parameters: Set the Time That Travelers Can Wait Untill to be Matched to a Driver 

```python
params.times.patience = 600 #in second
```

#### Parameters: Set the Value of Time for Travelers

```python
params.VoT = 10.63 # Each hour is equivalent to this amount of Euro
```

#### Parameters: Set the Path for the Demand (With No Demand Provided, MaaSSim Will generate the Demand)

```python
params.paths.requests = 'Amsterdam_requests.csv'
```

#### Parameters: Set the Seed to Generate a Specific Sample Out of Given Demand

```python
params.seed = 1
```

#### Parameters: Set the Min and Max Distances to Filter the Ride Requests

```python
params.dist_threshold = 20000 #meter
params.dist_threshold_min = 3000 #meter
```

#### Parameters: Set True if You Want Traveler/Driver Participation Decision to be Probabilistic (Otherwise its Deterministic)

```python
params.d2d.probabilistic = True
```

#### Parameters: Set True if You Want Travelers/Drivers to be Heterogeneous (Specific Reservation Wage for Each Driver and Specific Value of Time for Each Traveler) 

```python
params.d2d.heterogeneous = False
params.proportional_std = 0.1 # proportional standard deviation [0.1, 0.2, 0.3] for both VoT & RW distributions
```

#### Parameters: Set the Initial Utility Close to Zero (Zero itself will Raise and Error) 

```python
params.d2d.ini_att = 0.01
params.d2d.Eini_att = 0.01
```

#### Parameters: Set the Betas for Traveler/Driver Utility Functions

```python
params.d2d.B_income = 1 # The weight of income in driver utility equation
params.d2d.B_inveh_time = 1 # The weight of in-vehicle time in traveler utility equation
params.d2d.B_exp_time = 2 # The weight of waiting time in traveler utility equation
```

#### Parameters: Set the Trip Fare for Public Transportation for Travelers

```python
params.PT_fare = 1 # in Euro
```

#### Parameters: Set the Public Transport Speed for Travelers

```python
params.PT_avg_speed = 6 # in m/s
```

#### Parameters: Set the Number of Platforms in the Market

```python
params.nPM = 1 # platform id starts from 1
```

#### Parameters: Set the Platform Fares (Use Platform Index to Define Specific Fares for Each Platform)

```python
params.platforms.base_fare = 1.4 # in Euro 
params.platforms.fare = float(1.2)# in Euro/km (euro/km #http://taxihowmuch.com/)
params.platforms.min_fare = float(4) # in Euro
```

#### Parameters: Set the Platform Levers

```python
params.platforms.comm_rate = float(0.20) # 20%
params.platforms.discount = float(0.0) # 0%
params.platforms.incentive = float(0.0) # 0%
params.platforms.daily_marketing = False
```

#### Parameters: Set the Weight for Each Utility Component

```python
params.d2d.B_Experience = 0.70
params.d2d.B_WOM = 0.20
params.d2d.B_Marketing = 0.10
```

#### Parameters: Set the Diffusion Speed of Word-of-Mouth and Marketing Among the Agents

```python
params.d2d.diffusion_speed = 10/100 # 100/100 is the max speed
```

#### Parameters: Set the Learning Degree for Travelers/Drivers (with Greater Values Agents Will Learn Faster)

```python
params.d2d.learning_d = 1 
params.d2d.adj_s = 2 # Complementary component of learning
```

#### Parameters: Set the Fuel Cost for Drivers Which is Used to Calculate Their Profit

```python
params.d2d.fuel_cost = 0.25 #in Euro/km (https://help.ubercarshare.com/hc/en-gb/articles/360022882131-Choosing-your-car-s-distance-rate)  
```

#### Parameters: Set True if You Have Stop Creteria for the Simulation Or Assertation Function 

```python
params.f_stop_crit = False
params.assert_me = False
```

#### Parameters: Set the Mu for the Logit Function Used to Model Agents' Participation Decision 

```python
params.d2d.m = 5
```

#### Parameters: Set the Number of Drivers (nV) and Travelers (nP) 

```python
params.nP = 200
params.nV = 20
```

#### Parameters: Set the Number of Days for Simulation and Duration of Each Simulation Day

```python
params.nD = 5
params.simTime = 4 
```

#### Run the Simulation

```python
sim = MaaSSim.simulators.simulate(params=params, f_driver_out=S_driver_opt_out, f_trav_out=S_traveller_opt_out,
                                  kpi_veh=d2d_kpi_veh,kpi_pax=d2d_kpi_pax, logger_level=logging.WARNING)
```
