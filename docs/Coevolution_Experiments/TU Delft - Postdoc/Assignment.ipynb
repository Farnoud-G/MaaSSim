#### MaaSSim

<div style="text-align: justify;">
MaaSSim is a simulation environment that enables detailed interactions between travelers and drivers on a given road network graph. On the supply side, it explicitly represents individual vehicles and their movements across time and space. On the demand side, it captures precise trip requests, including exact request times and destinations defined at the graph node level. The simulation outputs detailed sequences of space-time locations and statuses for both vehicles and travelers, which are further synthesized into agent-level and system-wide key performance indicators (KPIs) for in-depth analysis.
</div>
